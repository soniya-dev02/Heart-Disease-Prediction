```python
#Data Labeling

x=df.drop('HeartDisease',axis=1)
y=df['HeartDisease']
```

```python
x.columns
```

```python
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
#Model training
#logistic regression

#1. import algorithm
from sklearn.linear_model import LogisticRegression
#2. object for class
#model
lr=LogisticRegression()
```

```python
#Model training
lr.fit(x_train,y_train)
```

```python
#testing---x_test
lr_result=lr.predict(x_test)
print(lr_result)
```

```python
#model evaluation

from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

score_lr=accuracy_score(y_test,lr_result)
print("Accuracy Score for LR Model:",score_lr)
```

```python
#classification report

report=classification_report(y_test,lr_result)
print("classification report:\n")
print(report)
```

```python
print(y_train.value_counts())
```

```python
#confusion matrix

import seaborn as sns
import matplotlib.pyplot as plt

cm=confusion_matrix(y_test,lr_result)
print(cm)

plt.figure(figsize=(8,6),dpi=100)
plt.title("Logistic Regression Confusion Matrix")

sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.show()
```

```python
#over sampling method of balancing

from imblearn.over_sampling import SMOTE
smt=SMOTE()
x_train_sampled,y_train_sampled=smt.fit_resample(x_train,y_train)
```

```python
#Model training
#logistic regression

#1. import algorithm
from sklearn.linear_model import LogisticRegression
#2. object for class
#model
lr=LogisticRegression(class_weight='balanced')

lr.fit(x_train_sampled,y_train_sampled)
```

```python
lr_result=lr.predict(x_test)
print(lr_result)
```

```python
#confusion matrix

import seaborn as sns
import matplotlib.pyplot as plt

cm=confusion_matrix(y_test,lr_result)
print(cm)

plt.figure(figsize=(8,6),dpi=100)
plt.title("Logistic Regression Confusion Matrix")

sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.show()
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

score_lr=accuracy_score(y_test,lr_result)
print("Accuracy Score for LR Model:",score_lr)
```

```python
#decision tree classifier
from sklearn.tree import DecisionTreeClassifier

#model building

dt=DecisionTreeClassifier()

dt.fit(x_train_sampled,y_train_sampled)

#model testing---x_data
result_dt=dt.predict(x_test)
print(result_dt)

#model evaluation
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

score_dt=accuracy_score(y_test,result_dt)
print("Accuracy score for dt model:",score_dt)

#classification report
report_dt=classification_report(y_test,result_dt)
print("classification report:\n")
print(report_dt)

import seaborn as sns
import matplotlib.pyplot as plt

cm=confusion_matrix(y_test,result_dt)
print(cm)

plt.figure(figsize=(8,6),dpi=100)
plt.title("Decision Tree Confusion Matrix")

sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

#model building
rf=RandomForestClassifier()
rf.fit(x_train_sampled,y_train_sampled)

#model testing---x_test data
result_rf=rf.predict(x_test)
print(result_rf)

#model evaluation
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report


score_rf=accuracy_score(y_test,result_rf)
print("Accuracy score for rf model:",score_dt)

#classification report
report_rf=classification_report(y_test,result_rf)
print("classification report:\n")
print(report_rf)

import seaborn as sns
import matplotlib.pyplot as plt

cm=confusion_matrix(y_test,result_rf)
print(cm)

plt.figure(figsize=(8,6),dpi=100)
plt.title("RandomForest Confusion Matrix")

sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.show()
```

```python
#support vector machine
from sklearn.svm import SVC

#model building
svc=SVC(class_weight='balanced')

#model training
svc.fit(x_train_sampled,y_train_sampled)

#model testing----x_test data
result_svc=svc.predict(x_test)
print(result_svc)

#model evaluation
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

score_svc=accuracy_score(y_test,result_svc)
print("Accuracy score for SVM model:",score_svc)

#classification report
report_svc=classification_report(y_test,result_svc)
print("classification report:\n")
print(report_svc)

import seaborn as sns
import matplotlib.pyplot as plt

cm=confusion_matrix(y_test,result_svc)
print(cm)

plt.figure(figsize=(8,6),dpi=100)
plt.title("SVM Confusion Matrix")

sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.show()

```

```python
#xg boost classifier

from xgboost import XGBClassifier

#model building
xgb=XGBClassifier()

#model training
xgb.fit(x_train_sampled,y_train_sampled)

#model testing
result_xgb=xgb.predict(x_test)
print(result_xgb)

#model evaluation
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

score_xgb=accuracy_score(y_test,result_xgb)
print("Accuracy score for XGBoost model:",score_xgb)

#classification report
report_xgb=classification_report(y_test,result_xgb)
print("classification report:\n")
print(report_xgb)

import seaborn as sns
import matplotlib.pyplot as plt

cm=confusion_matrix(y_test,result_xgb)
print(cm)

plt.figure(figsize=(8,6),dpi=100)
plt.title("XGBoost Confusion Matrix")

sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.show()
```
